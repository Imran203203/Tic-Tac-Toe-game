```python
###Funtion to print out a board.Setting our board as a list ,where each index 1-9 corresponds with a number on a number pad,so we 
###get a 3 by 3 board representation.
from IPython.display import clear_output
def display_board(board):
    clear_output()       #This only works in jupyter!This will clear the history of previous boards displayed if function is 
                        # called multiple times in same cell.
    
    print('  |   |')
    print(' '+board[7]+'| '+board[8]+' | '+board[9])
    print('  |   |')
    print('-----------')
    print('  |   |')
    print(' '+board[4]+'| '+board[5]+' | '+board[6])
    print('  |   |')
    print('-----------')
    print('  |   |')
    print(' '+board[1]+'| '+board[2]+' | '+board[3])
    print('  |   |')
    
    
```

```python
#Function to take in a player input and assign their marker as 'X' or 'O'
def player_input():
    marker='wrong'
    while marker!='X' and marker!='O':
        marker=input('Player1: Choose your marker- X or O').upper()
    if marker=='X':
        
        return ('X','O')
    else:
        return ('O','X')
```

```python
#Function that takes in the board list object, a marker ('X' or 'O') and a desired position(number 1-9) and assigns it to the 
#board
def place_marker(board,marker,position):
    board[position]=marker
```

```python
#Function that takes in a board and checks to see if someone has won.
def win_check(board,mark):
    return ((board[7]==board[8]==board[9]==mark) or
           (board[4]==board[5]==board[6]==mark) or
           (board[1]==board[2]==board[3]==mark) or
           (board[7]==board[4]==board[1]==mark) or
           (board[8]==board[5]==board[2]==mark) or
           (board[9]==board[6]==board[3]==mark) or
           (board[1]==board[5]==board[9]==mark) or
           (board[3]==board[5]==board[7]==mark))
```

```python
#Function to decide which player goes first and return a string which player went first.
import random
def choose_first():
    flip=random.randint(0,1)
    if flip==0:
        return 'Player 1'
    else:
        return 'Player 2'
```

```python
#Function to check if space is freely available on the board.
def space_check(board,position):
    return board[position]==' '
```

```python
#Function to check if the board is full and returns True if Full,False otherwise.
def full_board_check(board):
    for i in range(1,10):
        if space_check(board,i):
            return False
    return True
```

```python
#Function that asks for a player's next position (as a number 1-9) and check if it is a free position.If it is,then return the
#position for later use
def player_choice(board):
    position=0
    while position not in [1,2,3,4,5,6,7,8,9] or not space_check(board,position):
        position=int(input('Choose your next position:(1-9)'))

    return position
```

```python
#Function that asks the player if they want to play again and returns a boolean True if they do want to play again.
def replay():
    return input('Do you want to play again?Enter Yes or NO:').lower().startswith('y')
```

```python
# print('Welcome to Tic Tac Toe!')
while True:
    #Reset the board
    board=[' ']*10
    Player1_mark,Player2_mark=player_input()
    turn=choose_first()
    print(turn+' will go first')
    play_game = input('Are you ready to play? Enter Yes or No.')
    
    if play_game.lower()[0] == 'y':
        game_on = True
    else:
        game_on = False

    while game_on:
        if turn=='Player 1':
            #Player1's turn
            
            display_board(board)
            choice=player_choice(board)
            place_marker(board,Player1_mark,choice)
            if win_check(board,Player1_mark):
                display_board(board)
                print('Congratulations! Player 1,you have won')
                game_on=False
            else:
                if full_board_check(board):
                    display_board(board)
                    print('Game Tie!')
                    game_on=False
                else:
                    turn='Player 2'
        else:
            #Player2's turn
            display_board(board)
            choice=player_choice(board)
            place_marker(board,Player2_mark,choice)
            if win_check(board,Player2_mark):
                display_board(board)
                print('Congratulations!Player 2,you have won')
                game_on=False
            else:
                if full_board_check(board):
                    display_board(board)
                    print('Game Tie!')
                    game_on=False
                else:
                    turn='Player 1'
        
    if not replay():
        break
```
